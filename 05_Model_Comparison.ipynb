# Project Overview

Credit card fraud detection is a highly imbalanced binary classification problem where fraudulent transactions represent only a very small fraction of all transactions. In such scenarios, relying solely on accuracy is misleading because a model can achieve very high accuracy while failing to detect fraudulent transactions.

The primary objective of this project was to build a robust fraud detection system capable of identifying fraudulent transactions while maintaining a balance between:

- High Fraud Detection Rate (Recall)
- Low False Positive Rate (Precision)
- Strong Overall Classification Performance (F1-Score)
- Excellent Ranking Ability (ROC-AUC)
- High Performance on Imbalanced Data (PR-AUC)
- Model Interpretability and Reliability

To achieve this, three state-of-the-art Gradient Boosting algorithms were developed, optimized, and evaluated under the same preprocessing pipeline and evaluation strategy.

# Threshold Optimization Summary

Threshold optimization was performed for every model to analyze the trade-off between Precision and Recall.

### Key Findings

### LightGBM

- Highest Recall among all models.
- Produced a large number of False Positives.
- Less suitable for real-world deployment where unnecessary fraud alerts increase operational costs.

---

### XGBoost

- Good balance between Precision and Recall.
- Stable performance across thresholds.
- Strong alternative model with consistent predictions.

---

### CatBoost

Two different behaviors were observed.

**Balanced Class Weights**

- High Recall (91.26%)
- Good Precision (73.73%)
- Excellent ROC-AUC
- Highest PR-AUC
- Better ability to capture fraudulent transactions.

**Optuna Tuned**

- Extremely High Precision (96.83%)
- Very Low False Positives
- Lower Recall compared to Balanced model.

This indicates that the Optuna model became more conservative, whereas the Balanced model remained more effective at detecting fraud.

# Why CatBoost Was Selected

Although multiple models produced excellent results, **CatBoost with Balanced Class Weights** was selected as the final model.

This decision was based on overall business performance rather than a single evaluation metric.

### 1. Highest ROC-AUC

CatBoost Balanced achieved

**ROC-AUC = 0.998731**

which was the highest among all evaluated models.

This indicates excellent ranking capability and strong discrimination between legitimate and fraudulent transactions.

---

### 2. Highest PR-AUC

Fraud detection is an extremely imbalanced classification problem.

For such datasets, PR-AUC is often more informative than ROC-AUC.

CatBoost Balanced achieved

**PR-AUC = 0.939017**

which was the highest among all models.

This demonstrates superior performance on the minority fraud class.

---

### 3. Excellent Fraud Detection

Recall measures the percentage of fraudulent transactions successfully detected.

CatBoost Balanced achieved

**Recall = 91.26%**

meaning that the model successfully identified over 91% of fraudulent transactions.

For fraud detection systems, missing fraud (False Negatives) is generally more expensive than reviewing additional alerts.

---

### 4. Better Business Trade-Off

Compared to the Optuna model:

| Metric | Balanced | Optuna |
|---------|---------:|--------:|
| Precision | 73.73% | 96.83% |
| Recall | **91.26%** | 82.29% |
| False Positives | 145 | **12** |
| False Negatives | **39** | 79 |

The Optuna model reduced False Positives significantly but nearly doubled the number of missed fraudulent transactions.

Since the primary objective of fraud detection is to identify as many fraudulent transactions as possible, the Balanced model provides a more practical trade-off between fraud detection and operational cost.

---

### 5. Strong Model Interpretability

SHAP analysis consistently identified the most influential features, including:

- Category Amount Z-Score
- Transaction Amount
- Business Hours
- Time-Based Features
- Category Statistics

These results aligned closely with Gain Feature Importance, increasing confidence in the model's decision-making process.

# Limitations

Although the selected model demonstrated excellent predictive performance, several limitations remain:

- The dataset represents historical fraud patterns and may not capture newly emerging fraud strategies.
- The model was trained using offline batch data rather than real-time transaction streams.
- Model performance may gradually decline over time due to concept drift.
- Additional behavioral, geographical, and device-based features could further improve fraud detection performance.
